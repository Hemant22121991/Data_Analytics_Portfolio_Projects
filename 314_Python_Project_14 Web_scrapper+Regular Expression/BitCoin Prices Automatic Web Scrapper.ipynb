```python
from bs4 import BeautifulSoup
import requests
import os
import pandas as pd
from datetime import datetime
import time
```

```python
def automated_crpto_pull():

    url = 'https://coinmarketcap.com/currencies/bitcoin/'
    
    page = requests.get(url)
    
    soup = BeautifulSoup(page.text, 'html')
    
    crypto_name = soup.find('span',class_="sc-65e7f566-0 lsTl").text
    
    crypto_price = soup.find('span', class_="sc-65e7f566-0 WXGwg base-text").text
    
    final_price = crypto_price.replace('$','')
    
    date_time = datetime.now()
    
    dict = {'crypto Name': crypto_name,
            'Price':final_price,
           'TimeStamp': date_time}
    
    df = pd.DataFrame([dict])
    
    if os.path.exists('Bitcoin_price.csv'):
        df.to_csv('Bitcoin_price.csv',mode = 'a',header= False)
    else:
        df.to_csv('Bitcoin_price.csv')
    print(df)
```

```python
while True:
    automated_crpto_pull()
    time.sleep(3600)
```
